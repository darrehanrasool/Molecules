# 04 - Baseline Spectral Retrieval Models**Enveda CASMI 2026: Molecule ID From Mass Spectra**  *Centre of Artificial Intelligence, IUST Kashmir*Demonstrating precursor-constrained candidate filtering, spectral cosine scoring, and InChIKey14 deduplication.

```python
from molecules.models.retrieval import PrecursorRetrievalPredictor
from molecules.postprocessing.ranker import CandidateRanker
from molecules.data.loader import MassSpectrumRecord

# Mock candidate catalog
catalog = pd.DataFrame([
    {"smiles": "O=C1C=C(c2ccccc2)Oc3cc(O)cc(O)c13", "neutral_mass": 254.0579, "peaks": "105.03:0.42 153.01:0.85 254.05:1.0"},
    {"smiles": "OCC1OC(O)C(O)C(O)C1O", "neutral_mass": 180.0634, "peaks": "60.02:0.5 73.03:1.0"},
])

spec = MassSpectrumRecord(
    molecule_id="query_01",
    precursor_mz=255.0652,
    adduct="[M-H]-",
    peaks=np.array([[105.03, 0.42], [153.01, 0.85], [254.05, 1.0]]),
    smiles="O=C1C=C(c2ccccc2)Oc3cc(O)cc(O)c13"
)

retriever = PrecursorRetrievalPredictor(candidate_catalog=catalog, ppm_tolerance=15.0)
scored = retriever.predict_candidates(spec, top_k=25)
ranker = CandidateRanker(max_candidates=25)
top_smiles = ranker.rank_and_deduplicate(scored)
print("Top-ranked predictions:")
for r, sm in enumerate(top_smiles, 1):
    print(f"  Rank {r}: {sm}")
```
