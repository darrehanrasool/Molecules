## 2. Load Sample Spectrum and Visualize Peaks

```python
sample_peaks = np.array([
    [105.033, 0.42],
    [153.018, 0.85],
    [209.045, 0.23],
    [254.057, 1.00], # Base peak
    [255.065, 0.15]
])

plt.figure(figsize=(10, 4), dpi=150)
for mz, intensity in sample_peaks:
    plt.vlines(x=mz, ymin=0, ymax=intensity, color='#1f77b4', lw=2)
plt.title("Sample MS/MS Spectrum (m/z vs. Relative Intensity)")
plt.xlabel("Fragment m/z (Da)")
plt.ylabel("Relative Abundance")
plt.ylim(0, 1.1)
plt.grid(True, alpha=0.3)
plt.show()
```
