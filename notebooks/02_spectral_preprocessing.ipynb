```python
from molecules.data.preprocessor import SpectralPreprocessor, compute_neutral_losses
from molecules.data.spectral_features import spectrum_to_dense_bins

preprocessor = SpectralPreprocessor(min_intensity_rel=0.005, normalize=True)
raw_peaks = np.array([
    [100.0, 10.0],
    [200.0, 50.0], # Base peak
    [300.0, 0.05], # Noise below 0.5%
])
cleaned = preprocessor.process(raw_peaks)
print("Cleaned Peaks:")
print(cleaned)
```

## Compute Dense Feature Vector

```python
dense_vec = spectrum_to_dense_bins(cleaned, min_mz=10.0, max_mz=500.0, bin_size=1.0)
print(f"Dense Binned Vector Shape: {dense_vec.shape}")
print(f"Non-zero bins count      : {np.count_nonzero(dense_vec)}")
```
